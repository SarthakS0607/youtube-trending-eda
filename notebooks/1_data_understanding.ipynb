```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

df = pd.read_csv("../data/INvideos.csv")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
print("Exact duplicate rows:", df.duplicated().sum())   # Check for exact number of duplicate rows 
```

```python
print("Unique video IDs:", df["video_id"].nunique())
print("Unique channels:", df["channel_title"].nunique())
print("Unique categories:", df["category_id"].nunique())
print("Unique trending dates:", df["trending_date"].nunique())
```

```python
df.sort_values("views", ascending=False)[
    ["video_id", "title", "channel_title", "views", "likes", "comment_count"]
].head(10)
```

```python
high_view_videos = df[df["views"] > 1_000_000]

print("Videos with more than 1 million views:", len(high_view_videos))

high_view_videos[
    ["title", "channel_title", "views", "likes", "comment_count"]
].head(10)
```

```python
missing_values = df.isna().sum().sort_values(ascending=False)

missing_values[missing_values > 0]
```
